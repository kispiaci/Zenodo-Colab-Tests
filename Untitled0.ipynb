```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import gaussian_filter

# --- MÉRET ---
N = 400  # Kicsit nagyobb felbontás

# --- PARAMÉTEREK ---
# A Kuramoto-alapú ZPF dinamikához
dt = 0.1
steps = 600
K = 1.8         # Erősebb szinkronizáció a szálakhoz
sigma_long = 12 # Globális rend (mély hangok)
sigma_short = 2 # Lokális sűrűsödés (magas hangok)
alpha = 0.5     # Nem-lokális visszacsatolás erőssége

# --- KEZDŐÁLLAPOT ---
# Véletlen fázisok + egy kis alapstruktúra (mint a CMB fluktuációi)
phi = np.random.rand(N, N) * 2 * np.pi

def get_density(p):
    # A sűrűség a fázisok koherenciájából adódik (ZPF kondenzáció)
    # Ahol a fázisok hasonlók, ott "anyag" keletkezik
    cos_p = np.cos(p)
    sin_p = np.sin(p)
    # Lokális koherencia mérése simítással
    coherence = np.sqrt(gaussian_filter(cos_p, sigma=sigma_short)**2 +
                        gaussian_filter(sin_p, sigma=sigma_short)**2)
    return coherence

# --- FEJLŐDÉS ---
plt.figure(figsize=(10, 8))

for step in range(steps):
    # 1. Globális fázisátlag (a "karmester" hangja)
    global_phase = np.arctan2(np.mean(np.sin(phi)), np.mean(np.cos(phi)))

    # 2. Kuramoto-szinkronizáció a globális és lokális mezővel
    sync_term = K * np.sin(phi - global_phase)

    # 3. Nem-lokális gátlás (ez csinálja az üres völgyeket /voids/)
    phi_blur = gaussian_filter(phi, sigma=sigma_long)
    nonlocal_term = alpha * np.sin(phi - phi_blur)

    # 4. Fázisfrissítés
    phi += dt * (sync_term - nonlocal_term)

    # Egy kis diffúzió, hogy ne legyen zajos
    phi = gaussian_filter(phi, sigma=0.5)

    if step % 200 == 0:
        print(f"Lépés: {step}...")

# --- VIZUALIZÁCIÓ (Sűrűség leképezése) ---
# Nem a fázist nézzük, hanem a belőle eredő anyagkoncentrációt
density = get_density(phi)
# Kontraszt növelése a filamentumok kiemeléséhez
density = (density - np.min(density)) / (np.max(density) - np.min(density))

plt.imshow(density**2, cmap='magma', origin='lower')
plt.title("ZPF-Kuramoto Kozmikus Háló Szimuláció")
plt.axis('off')
plt.show()
```
